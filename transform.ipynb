```python
import pandas as pd
```

```python
data = pd.read_excel("mmm_104_weeks_dataset.xlsx", sheet_name="MMM_Data")
```

```python
youtubeDF = data[['Week','YouTube','Sales']]
```

```python
youtubeDF
```

```python
youtubeDF['YouTube'].median() 
```

```python
youtubeDF['Youtube_Standardized'] = youtubeDF['YouTube']/ youtubeDF['YouTube'].median() 
```

```python
youtubeDF
```

```python
import numpy as np

def adstock_series(series, decay):
    adstocked = np.zeros(len(series))
    adstocked[0] = series.iloc[0]
    
    for t in range(1, len(series)):
        adstocked[t] = series.iloc[t] + decay * adstocked[t-1]
    
    return adstocked

# Apply
decay = 0.5
youtubeDF['Youtube_Adstock'] = adstock_series(youtubeDF['Youtube_Standardized'], decay)
```

```python
import numpy as np

def hill_transform(x, alpha, gamma):
    return (x ** alpha) / (x ** alpha + gamma ** alpha)
```

```python
alpha = 1.5   # slope
gamma = 200   # half-saturation

youtubeDF['Youtube_Hill'] = hill_transform(
    youtubeDF['Youtube_Adstock'],
    alpha,
    gamma
)
```
